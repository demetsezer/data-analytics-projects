**Import libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

**Getting data**

```python
data = pd.read_csv('Spotify_Youtube.csv')
data
```

**Dropping unnecessary columns**

```python
data.columns
```

```python
data.drop(['Unnamed: 0', 'Url_spotify', 'Uri', 'Url_youtube', 'Description','Title'], axis=1, inplace=True)
data
```

```python
data.info()
```

**Dropping nulls and duplicates**

```python
data.isna().sum()
```

```python
data['Likes'] = data['Likes'].fillna(0)
data['Comments'] = data['Comments'].fillna(0)
data.dropna(inplace=True)
data.info()
```

```python
data.dropna(inplace=True)
data.info()
```

```python
data.drop_duplicates(inplace=True)
data.info()
```

```python
data.describe()
```

**Album Type**

```python
album_type = data['Album_type'].value_counts()
album_type
```

```python
plt.figure(figsize=(10,5))
album_type.plot(kind='pie', autopct='%1.1f%%')
plt.title('Album Type')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

**Artist Analysis**

```python
artist_group = data.groupby('Artist').agg({'Views':'sum', 'Stream':'sum'})
artist_group['Total'] = artist_group['Views'] + artist_group['Stream']
artist_group_sorted = artist_group.sort_values('Total', ascending=False)
artist_group.reset_index(inplace=True)
artist_group_sorted.reset_index(inplace=True)
artist_group
```

```python
artist_group_sorted
```

```python
top_10_artist = artist_group_sorted.head(10)
plt.figure(figsize=(10,5))
plt.bar(top_10_artist['Artist'], top_10_artist['Total'])
plt.title('Total Views and Streams by Artist')
plt.xlabel('Artist')
plt.ylabel('Total')
plt.xticks(rotation=90)
plt.show()
```

```python
youtube_columns = ['Artist', 'Views']
top_10_artist_youtube = artist_group_sorted[youtube_columns].head(10)
top_10_artist_youtube = top_10_artist_youtube.sort_values('Views', ascending=False)
plt.figure(figsize=(10,5))
plt.bar(top_10_artist_youtube['Artist'], top_10_artist_youtube['Views'])
plt.title('Total Youtube Views by Artist')
plt.xlabel('Artist')
plt.ylabel('Youtube Views')
plt.xticks(rotation=90)
plt.show()
```

```python
spotify_columns = ['Artist', 'Stream']
top_10_artist_spotify = artist_group_sorted[spotify_columns].head(10)
top_10_artist_spotify = top_10_artist_spotify.sort_values('Stream', ascending=False)
plt.figure(figsize=(10,5))
plt.bar(top_10_artist_spotify['Artist'], top_10_artist_spotify['Stream'])
plt.title('Total Spotify Streams by Artist')
plt.xlabel('Artist')
plt.ylabel('Spotify Streams')
plt.xticks(rotation=90)
plt.show()
```

```python
youtube_columns = ['Track', 'Views']
top_10_song_youtube = song_group_sorted[youtube_columns].head(10)
top_10_song_youtube = top_10_song_youtube.sort_values('Views', ascending=False)
plt.figure(figsize=(10,5))
plt.bar(top_10_song_youtube['Track'], top_10_song_youtube['Views'])
plt.title('Total Youtube Views by Track')
plt.xlabel('Track')
plt.ylabel('Youtube Views')
plt.xticks(rotation=90)
plt.show()
```

```python
top_10_liked_songs = likes_sorted.head(10)
plt.figure(figsize=(10,5))
plt.bar(top_10_liked_songs['Track'], top_10_liked_songs['Likes'])
plt.title('Total Likes by Track')
plt.xlabel('Track')
plt.ylabel('Likes')
plt.xticks(rotation=90)
plt.show()
```

**Channel Analysis**

```python
channel_group = data.groupby('Channel').agg({'Views':'sum', 'Stream':'sum'})
channel_group['Total'] = channel_group['Views'] + channel_group['Stream']
channel_group_sorted = channel_group.sort_values('Total', ascending=False)
channel_group.reset_index(inplace=True)
channel_group_sorted.reset_index(inplace=True)
top_10_channel = channel_group_sorted.head(10)
top_10_channel
```

**Dancebility Analysis**

```python
technical_features = data.groupby('Track').agg({'Danceability':'mean', 'Acousticness':'mean', 'Energy':'mean', 'Instrumentalness':'mean', 
                                          'Liveness':'mean', 'Valence':'mean', 'Tempo':'mean'})

technical_features_sorted = technical_features.sort_values('Danceability', ascending=False)
technical_features_sorted.reset_index(inplace=True)
top_10_danceability = technical_features_sorted.head(10)
```

**Duration Analysis**

```python
data['Duration_s'] = data['Duration_ms'] / 1000
data['Duration_minute'] = data['Duration_s'] / 60
data['Duration_minute'] = data['Duration_minute'].round(2)
duration = data.groupby('Duration_minute').agg({'Track':'count'}).rename(columns={'Track':'Count'})
duration_sorted = duration.sort_values('Count', ascending=False)
duration_sorted.reset_index(inplace=True)
top_10_duration = duration_sorted.head(10)
top_10_duration
```

```python
duration_sorted_order = duration_sorted.sort_values('Duration_minute', ascending=True)
duration_sorted_order.reset_index(inplace=True)
duration_sorted_order
```

```python
plt.figure(figsize=(10,5))
plt.bar(duration_sorted_order['Duration_minute'], duration_sorted_order['Count'])
plt.title('Total Songs by Duration')
plt.xlabel('Duration (minute)')
plt.ylabel('Count')
plt.show()
```

```python
duration_sorted_order = duration_sorted_order[duration_sorted_order['Duration_minute'] < 10]
plt.figure(figsize=(10,5))
plt.bar(duration_sorted_order['Duration_minute'], duration_sorted_order['Count'])
plt.title('Total Songs by Duration')
plt.xlabel('Duration (minute)')
plt.ylabel('Count')
plt.show()
```

**Total watch - Likes**

```python
data_likes_watch = data.groupby('Track').agg({'Likes':'sum', 'Stream':'sum', 'Views':'sum'})
data_likes_watch['Total'] = data_likes_watch['Stream'] + data_likes_watch['Views']
data_likes_watch['Ratio'] = data_likes_watch['Likes'] / data_likes_watch['Total']
data_likes_watch_sorted = data_likes_watch.sort_values('Ratio', ascending=False)
data_likes_watch_sorted.reset_index(inplace=True)
top_10_ratio = data_likes_watch_sorted.head(10)
top_10_ratio
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_10_ratio['Track'], top_10_ratio['Ratio'])
plt.title('Total Likes to Watch Ratio by Track')
plt.xlabel('Track')
plt.ylabel('Ratio')
plt.xticks(rotation=90)
plt.show()
```

```python
artist_likes_watch = data.groupby('Artist').agg({'Likes':'sum', 'Stream':'sum', 'Views':'sum'})
artist_likes_watch['Total'] = artist_likes_watch['Stream'] + artist_likes_watch['Views']
artist_likes_watch['Ratio'] = artist_likes_watch['Likes'] / artist_likes_watch['Total']
artist_likes_watch_sorted = artist_likes_watch.sort_values('Ratio', ascending=False)
artist_likes_watch_sorted.reset_index(inplace=True)
top_10_artist_ratio = artist_likes_watch_sorted.head(10)
top_10_artist_ratio
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_10_artist_ratio['Artist'], top_10_artist_ratio['Ratio'])
plt.title('Total Likes to Watch Ratio by Artist')
plt.xlabel('Artist')
plt.ylabel('Ratio')
plt.xticks(rotation=90)
plt.show()
```
